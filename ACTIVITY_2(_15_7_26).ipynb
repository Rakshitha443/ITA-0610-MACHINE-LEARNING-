```python
import numpy as np


def sigmoid(z):
    """Sigmoid activation function."""
    return 1.0 / (1.0 + np.exp(-z))


def sigmoid_derivative(a):
    """Derivative of sigmoid function given activation output 'a'."""
    return a * (1.0 - a)


# =====================================================================
# ACTIVITY 1: Neural Network Representation (Mapping Student Data)
# =====================================================================
print("=== ACTIVITY 1: Neural Network Representation ===")

# Feature Encoding: High/Good = 1, Low/Poor = 0, Pass = 1, Fail = 0
student_data = [
    {"Attendance": 1, "Assignment": 1, "Exam": 1, "Result": 1},  # Pass
    {"Attendance": 0, "Assignment": 0, "Exam": 0, "Result": 0},  # Fail
]

print("Encoded Data Representation:")
for sample in student_data:
    features = [
        sample["Attendance"],
        sample["Assignment"],
        sample["Exam"],
    ]
    target = sample["Result"]
    print(
        f"  Inputs [x1={features[0]}, x2={features[1]}, x3={features[2]}] -> Target y={target}"
    )

print("Architecture: 3 Input Neurons -> 2 Hidden Neurons -> 1 Output Neuron\n")


# =====================================================================
# ACTIVITY 2: Perceptron Logic Gate Implementation (AND Gate)
# =====================================================================
print("=== ACTIVITY 2: Perceptron Learning for AND Gate ===")

# Dataset for AND Gate
X_and = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_and = np.array([0, 0, 0, 1])

# Initial parameters
weights = np.array([0.0, 0.0])
bias = -0.5
learning_rate = 0.1
epochs = 10

print(
    f"Initial State: Weights={weights}, Bias={bias}, Learning Rate={learning_rate}"
)

# Training loop using Perceptron Learning Rule
for epoch in range(1, epochs + 1):
    errors = 0
    for i in range(len(X_and)):
        z = np.dot(X_and[i], weights) + bias
        prediction = 1 if z >= 0 else 0
        error = y_and[i] - prediction

        if error != 0:
            weights += learning_rate * error * X_and[i]
            bias += learning_rate * error
            errors += 1

    if errors == 0:
        print(f"Converged successfully at Epoch {epoch}!")
        break

print(f"Trained Weights: {weights}, Trained Bias: {bias:.2f}")
print("Verification:")
for i in range(len(X_and)):
    z = np.dot(X_and[i], weights) + bias
    pred = 1 if z >= 0 else 0
    print(f"  Input: {X_and[i]} -> Output: {pred} (Expected: {y_and[i]})")
print()


# =====================================================================
# ACTIVITY 3: Perceptron Limitation (XOR Problem Demonstration)
# =====================================================================
print("=== ACTIVITY 3: Perceptron Limitation (XOR Problem) ===")

X_xor = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_xor = np.array([0, 1, 1, 0])

w_xor = np.array([0.0, 0.0])
b_xor = 0.0
converged = False

print("Attempting to train a single-layer Perceptron on XOR for 100 epochs...")
for epoch in range(100):
    errors = 0
    for i in range(len(X_xor)):
        z = np.dot(X_xor[i], w_xor) + b_xor
        pred = 1 if z >= 0 else 0
        err = y_xor[i] - pred
        if err != 0:
            w_xor += 0.1 * err * X_xor[i]
            b_xor += 0.1 * err
            errors += 1
    if errors == 0:
        converged = True
        break

print(
    f"Result: Converged = {converged} (Single-layer Perceptron failed to separate classes)"
)
print(
    "Explanation: XOR points are non-linearly separable; a Multilayer Perceptron is required.\n"
)


# =====================================================================
# ACTIVITY 4: Multilayer Perceptron (MLP) Forward Pass
# =====================================================================
print("=== ACTIVITY 4: MLP Forward Pass ===")

# Given inputs & weights
x = np.array([1, 0])
W_hidden = np.array(
    [[0.5, 0.3], [-0.5, 0.8]]
)  # Columns: weights for h1 and h2
W_output = np.array([0.7, -0.2])  # Weights from h1, h2 to output

# 1. Hidden Layer computation
z_hidden = np.dot(x, W_hidden)  # [z_h1, z_h2]
a_hidden = sigmoid(z_hidden)  # [a_h1, a_h2]

# 2. Output Layer computation
z_output = np.dot(a_hidden, W_output)
y_pred = sigmoid(z_output)

print(f"Inputs (x): {x}")
print(f"Hidden Pre-activations (z_h1, z_h2): {z_hidden}")
print(
    f"Hidden Activations     (a_h1, a_h2): [{a_hidden[0]:.4f}, {a_hidden[1]:.4f}]"
)
print(f"Output Pre-activation  (z_y)       : {z_output:.5f}")
print(f"Final Output Prediction (y_hat)    : {y_pred:.4f}\n")


# =====================================================================
# ACTIVITY 5: Backpropagation Weight Update
# =====================================================================
print("=== ACTIVITY 5: Backpropagation Weight Update ===")

target_y = 1.0
pred_y = 0.6
eta = 0.1
a_h1 = a_hidden[0]  # Activation feeding into weight = 0.6225
w_old = 0.7

# 1. Compute Mean Squared Error
loss = 0.5 * (target_y - pred_y) ** 2

# 2. Compute Output Error Delta: (y - y_hat) * σ'(z)
delta_y = (target_y - pred_y) * sigmoid_derivative(pred_y)

# 3. Gradient w.r.t weight: - delta_y * a_h
dE_dw = -delta_y * a_h1

# 4. Weight update via Gradient Descent
w_new = w_old - (eta * dE_dw)

print(f"Target Output (y)    : {target_y}")
print(f"Predicted Output (y_hat): {pred_y}")
print(f"Current Error (MSE)  : {loss:.4f}")
print(f"Output Delta (δ_y)   : {delta_y:.4f}")
print(f"Gradient (∂E/∂w)     : {dE_dw:.5f}")
print(f"Initial Weight (w_old): {w_old}")
print(f"Updated Weight (w_new): {w_new:.6f}")
```
