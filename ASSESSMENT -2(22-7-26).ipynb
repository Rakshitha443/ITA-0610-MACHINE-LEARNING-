```python
import numpy as np

# 1. Given Inputs and Parameters
x1 = 2.0   # Study Hours
x2 = 1.0   # Attendance

w1 = 0.6   # Weight for Study Hours
w2 = 0.4   # Weight for Attendance
b  = -0.5  # Bias

# 2. Net Input Calculation (z)
net_input = (w1 * x1) + (w2 * x2) + b

# 3. Step Activation Function
def step_function(z):
    return 1 if z >= 0 else 0

output = step_function(net_input)
result = "PASS" if output == 1 else "FAIL"

# 4. Display Step-by-Step Results
print("=" * 45)
print("     PERCEPTRON CLASSIFICATION RESULT     ")
print("=" * 45)
print(f"Study Hours (x1)     : {x1}")
print(f"Attendance (x2)      : {x2}")
print(f"Weight 1 (w1)        : {w1}")
print(f"Weight 2 (w2)        : {w2}")
print(f"Bias (b)             : {b}")
print("-" * 45)
print(f"Calculated Net Input (z) = ({w1} * {x1}) + ({w2} * {x2}) + ({b})")
print(f"Net Input (z)            = {net_input:.2f}")
print(f"Activation Output f(z)   = {output}")
print("-" * 45)
print(f"FINAL CONCLUSION         : Student will {result}")
print("=" * 45)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define grid range for plotting decision boundary
x1_vals = np.linspace(0, 3, 100)
# Decision Boundary Equation: w1*x1 + w2*x2 + b = 0  =>  x2 = (-w1*x1 - b) / w2
x2_boundary = (-w1 * x1_vals - b) / w2

plt.figure(figsize=(8, 6))

# Plot Decision Boundary
plt.plot(x1_vals, x2_boundary, 'r--', label='Decision Boundary (w1·x1 + w2·x2 + b = 0)')

# Fill Pass/Fail Regions
plt.fill_between(x1_vals, x2_boundary, 4, color='green', alpha=0.15, label='Pass Region (z >= 0)')
plt.fill_between(x1_vals, 0, x2_boundary, color='red', alpha=0.15, label='Fail Region (z < 0)')

# Plot Student's Data Point
plt.scatter([x1], [x2], color='blue', s=150, zorder=5, label=f'Student (x1={x1}, x2={x2}) -> {result}')

# Plot Styling
plt.title('Perceptron Classification Decision Boundary', fontsize=14, fontweight='bold')
plt.xlabel('Study Hours (x1)', fontsize=12)
plt.ylabel('Attendance (x2)', fontsize=12)
plt.xlim(0, 3)
plt.ylim(0, 3)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right')
plt.show()
```

```python
# =====================================================================
# NEURAL NETWORK: DELTA RULE WEIGHT UPDATE
# =====================================================================

# 1. Given Parameters
x = 2.0           # Input value
w_initial = 0.5   # Initial weight
eta = 0.1         # Learning rate (alpha)
target = 1.0      # Target output (t)
actual = 0.6      # Actual output (y)

# 2. Step 1: Compute the Output Error (E)
# Error = Target - Actual
error = target - actual

# 3. Step 2: Calculate Weight Update using Delta Rule (Δw)
# Delta Rule: Δw = η * (Target - Actual) * Input
delta_w = eta * error * x

# 4. Step 3: Compute New Updated Weight
w_new = w_initial + delta_w

# =====================================================================
# DISPLAY RESULTS
# =====================================================================
print("=" * 55)
print("     SINGLE NEURON WEIGHT UPDATE (DELTA RULE)     ")
print("=" * 55)
print(f"Input Value (x)          : {x}")
print(f"Initial Weight (w)       : {w_initial}")
print(f"Learning Rate (η)        : {eta}")
print(f"Target Output (t)        : {target}")
print(f"Actual Output (y)        : {actual}")
print("-" * 55)
print("STEP-BY-STEP CALCULATIONS:")
print(f"1. Error (e = t - y)      = {target} - {actual} = {error:.4f}")
print(f"2. Delta Weight (Δw)     = {eta} * {error:.4f} * {x} = {delta_w:.4f}")
print(f"3. New Weight (w_new)    = {w_initial} + {delta_w:.4f} = {w_new:.4f}")
print("=" * 55)
print(f"FINAL UPDATED WEIGHT     : {w_new:.4f}")
print("=" * 55)
```

```python
# =====================================================================
# GENETIC ALGORITHM: FITNESS, SELECTION, CROSSOVER & MUTATION
# =====================================================================

import numpy as np

# 1. INITIAL POPULATION & FITNESS COMPUTATION
# Function to maximize: f(x) = x^2
population = [2, 4, 6, 8]


def fitness(x):
    return x**2


fitness_scores = [fitness(x) for x in population]
total_fitness = sum(fitness_scores)

# Display Initial Population & Fitness
print("=" * 65)
print("     STEP 1: INITIAL POPULATION & FITNESS SCORES     ")
print("=" * 65)
print(f"{'Individual (x)':<15}{'Binary (4-bit)':<18}{'Fitness f(x)=x²':<18}{'Selection Prob (%)'}")
print("-" * 65)
for x, fit in zip(population, fitness_scores):
    binary_str = format(x, "04b")
    prob = (fit / total_fitness) * 100
    print(f"{x:<15}{binary_str:<18}{fit:<18}{prob:.2f}%")
print("=" * 65)


# 2. SELECT THE BEST CANDIDATES
# Selecting the top 2 individuals based on highest fitness scores
sorted_pop = [
    x for _, x in sorted(zip(fitness_scores, population), reverse=True)
]
selected_parents = sorted_pop[:2]  # Top 2: 8 and 6

print("\n" + "=" * 65)
print("     STEP 2: SELECTION OF BEST CANDIDATES     ")
print("=" * 65)
print(
    f"Selected Top Parents based on Fitness : {selected_parents[0]} (f={fitness(selected_parents[0])}) and {selected_parents[1]} (f={fitness(selected_parents[1])})"
)


# 3. CROSSOVER OPERATION (EXCHANGING LSB)
# Convert parents to 4-bit binary strings
p1_bin = format(selected_parents[0], "04b")  # 8 -> '1000'
p2_bin = format(selected_parents[1], "04b")  # 6 -> '0110'

# Exchange the LSB (last bit)
offspring1_bin = p1_bin[:-1] + p2_bin[-1]  # '100' + '0' = '1000' (8)
offspring2_bin = p2_bin[:-1] + p1_bin[-1]  # '011' + '0' = '0110' (6)

offspring1_val = int(offspring1_bin, 2)
offspring2_val = int(offspring2_bin, 2)

print("\n" + "=" * 65)
print("     STEP 3: LSB CROSSOVER OPERATION     ")
print("=" * 65)
print(
    f"Parent 1: {selected_parents[0]} ('{p1_bin[:-1]} | {p1_bin[-1]}')  -->  LSB = {p1_bin[-1]}"
)
print(
    f"Parent 2: {selected_parents[1]} ('{p2_bin[:-1]} | {p2_bin[-1]}')  -->  LSB = {p2_bin[-1]}"
)
print("-" * 65)
print(
    f"Offspring 1 (P1 head + P2 LSB): '{offspring1_bin}' = {offspring1_val} (f={fitness(offspring1_val)})"
)
print(
    f"Offspring 2 (P2 head + P1 LSB): '{offspring2_bin}' = {offspring2_val} (f={fitness(offspring2_val)})"
)
print("=" * 65)


# 4. EXPLANATION OF MUTATION
print("\n" + "=" * 65)
print("     STEP 4: ROLE OF MUTATION IN GENETIC ALGORITHMS     ")
print("=" * 65)
print("""
Role of Mutation:

1. Prevents Premature Convergence:
   Crossover only recombines existing genetic traits present in the population.
   If all individuals share identical bit positions, crossover cannot produce
   new search regions. Mutation randomly flips bits (e.g., 0 -> 1) to break out
   of local optima.

2. Maintains Genetic Diversity:
   Mutation restores lost or unrepresented alleles in the gene pool, ensuring
   the algorithm continues exploring unvisited regions of the search space.

3. Ensures Global Exploration:
   While Selection and Crossover focus on 'exploitation' (refining good solutions),
   Mutation acts as the primary driver for 'exploration' (discovering optimal ones).
""")
print("=" * 65)
```

```python
# =====================================================================
# OVERFITTING ANALYSIS & TRAINING CURVE VISUALIZATION
# =====================================================================

import matplotlib.pyplot as plt
import numpy as np

# 1. Simulate Epoch Data representing the scenario
epochs = np.arange(1, 21)

# Training accuracy increases steadily from 60% (0.60) to 95% (0.95)
train_acc = np.linspace(0.60, 0.95, 20)

# Validation accuracy increases from 58% (0.58) to 72% (0.72) at epoch 8,
# then degrades to 68% (0.68) in later epochs
val_acc = np.concatenate(
    [np.linspace(0.58, 0.72, 8), np.linspace(0.72, 0.68, 12)]
)

# =====================================================================
# PLOT TRAINING VS VALIDATION ACCURACY
# =====================================================================
plt.figure(figsize=(10, 6))
plt.plot(
    epochs,
    train_acc * 100,
    "b-o",
    linewidth=2,
    label="Training Accuracy (60% -> 95%)",
)
plt.plot(
    epochs,
    val_acc * 100,
    "r-o",
    linewidth=2,
    label="Validation Accuracy (58% -> 72% -> 68%)",
)

# Highlight point of overfitting onset
optimal_epoch = 8
plt.axvline(
    x=optimal_epoch,
    color="green",
    linestyle="--",
    linewidth=1.5,
    label=f"Overfitting Onset (Epoch {optimal_epoch})",
)

plt.title(
    "Training vs. Validation Accuracy Curve (Overfitting Analysis)",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Epochs", fontsize=12)
plt.ylabel("Accuracy (%)", fontsize=12)
plt.ylim(50, 100)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="lower right", fontsize=11)
plt.show()

# =====================================================================
# PRINT DETAILED ANALYSIS & THEORY
# =====================================================================
print("=" * 70)
print("             DEEP NEURAL NETWORK OVERFITTING ANALYSIS             ")
print("=" * 70)

print("""
1. IDENTIFIED ISSUE: OVERFITTING
----------------------------------------------------------------------
- Trend: From Epoch 1 to 8, both training and validation accuracies increase.
  However, after Epoch 8, training accuracy continues climbing to 95%, while
  validation accuracy drops from 72% to 68%.
- Analysis: This widening divergence between training performance and validation
  performance is a classic indicator of OVERFITTING.

2. WHY OVERFITTING OCCURS IN NEURAL NETWORKS:
----------------------------------------------------------------------
- Excessive Capacity / Complexity: Deep networks have millions of learnable
  parameters. Instead of learning generalizable underlying patterns, the model
  begins to "memorize" the training data, including its noise and outliers.
- Over-training (Too Many Epochs): Running training for too long allows the
  network to fit noise in the training set that does not exist in unseen data.

3. TWO EFFECTIVE TECHNIQUES TO ADDRESS OVERFITTING:
----------------------------------------------------------------------
a) Early Stopping:
   - Monitor validation accuracy/loss during training and halt execution when
     validation performance stops improving (e.g., at Epoch 8 in this setup)
     before degradation begins.
b) Regularization (e.g., Dropout / L2 Weight Decay):
   - Dropout randomly deactivates a fraction of neurons during training to prevent
     co-adaptation of features and force redundant representations.
   - L2 Regularization adds a penalty proportional to the square of weight values
     to the loss function, preventing weights from growing too large.

4. SIGNIFICANCE OF VALIDATION ACCURACY IN MODEL EVALUATION:
----------------------------------------------------------------------
- Generalization Capability: Training accuracy only measures how well a model
  fits seen data. Validation accuracy tests performance on unseen data, serving
  as an unbiased proxy for real-world deployment.
- Hyperparameter Tuning & Selection: Helps in selecting optimal hyperparameters
  (learning rate, architecture depth, epoch count) without contaminating test set data.
- Early Detection of Overfitting/Underfitting: Acts as a diagnostic signal to
  determine when to terminate training.
""")
print("=" * 70)
```

```python
# =====================================================================
# NEURAL NETWORK: DELTA LEARNING RULE WEIGHT UPDATE
# =====================================================================

# 1. Given Parameters
w_initial = 0.6  # Initial weight (w)
eta = 0.15  # Learning rate (η)
x = 2.5  # Input (x)
y_actual = 0.7  # Actual output (y)
t_target = 1.0  # Target output (t)

# 2. Step-by-Step Calculations
# Error Calculation: e = Target - Actual
error = t_target - y_actual

# Weight Change Calculation: Δw = η * (t - y) * x
delta_w = eta * error * x

# Updated Weight Calculation: w_new = w_initial + Δw
w_new = w_initial + delta_w

# =====================================================================
# DISPLAY COMPUTATIONS
# =====================================================================
print("=" * 60)
print("     DELTA LEARNING RULE WEIGHT UPDATE CALCULATION     ")
print("=" * 60)
print(f"Initial Weight (w)       : {w_initial}")
print(f"Learning Rate (η)        : {eta}")
print(f"Input (x)                : {x}")
print(f"Actual Output (y)        : {y_actual}")
print(f"Target Output (t)        : {t_target}")
print("-" * 60)
print("STEP-BY-STEP RESULTS:")
print(f"1. Error (e = t - y)    : {t_target} - {y_actual} = {error:.4f}")
print(
    f"2. Weight Update (Δw)    : η * e * x = {eta} * {error:.4f} * {x} = {delta_w:.4f}"
)
print(
    f"3. Updated Weight (w_new): w + Δw = {w_initial} + {delta_w:.4f} = {w_new:.4f}"
)
print("=" * 60)
```
