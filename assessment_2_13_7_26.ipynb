```python
# ==========================================
# CELL 1: Data Representation & Task A1 (Gini Index)
# ==========================================
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier, export_text

# Step 1: Data Representation
data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain', 'Rain', 'Overcast'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool', 'Cool', 'Mild'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal', 'Normal', 'Normal'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Weak', 'Strong', 'Strong'],
    'PlayTennis': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
print("--- Raw Dataset ---")
print(df)
print("\n" + "="*50 + "\n")

# Label Encoding for categorical values while maintaining mapping dictionary
label_encoders = {}
df_encoded = df.copy()

for col in df.columns:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df[col])
    label_encoders[col] = le

X = df_encoded.drop('PlayTennis', axis=1)
y = df_encoded['PlayTennis']

# Task A1: Train Model using Gini Heuristic
model_gini = DecisionTreeClassifier(criterion='gini', random_state=42)
model_gini.fit(X, y)

print("--- Task A1: Decision Tree Rules (Gini Index) ---")
rules_gini = export_text(model_gini, feature_names=list(X.columns))
print(rules_gini)

# Sample Prediction
# Encoding reference: Outlook (0:Overcast, 1:Rain, 2:Sunny), Humidity (0:High, 1:Normal)
sample = [[2, 1, 0, 1]]  # Example: Sunny, Hot, High, Weak
pred_gini = model_gini.predict(sample)
target_label = label_encoders['PlayTennis'].inverse_transform(pred_gini)[0]
print(f"Sample Input {sample} Prediction (Gini): {pred_gini[0]} -> '{target_label}'")
```
